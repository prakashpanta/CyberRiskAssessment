## 1. Accuracy Comparison Graph

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "/Users/prakashpanta/Desktop/Cyber_Security_Risk_Assessment/Outputs/Tables/All_Results.csv"
)

df.set_index("Experiment").plot(
    kind="bar",
    figsize=(10,6)
)

plt.title(
    "Accuracy Comparison Across Experiments"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.tight_layout()

plt.savefig(
    "/Users/prakashpanta/Desktop/Cyber_Security_Risk_Assessment/Outputs/Figures/Accuracy_Comparison.png",
    dpi=300
)

plt.show()
```

## 2. Original Vs Augmented Graph

```python
comparison = df[
    df["Experiment"].isin(
        ["E1_Original_Full", "E3_Augmented_Full"]
    )
]

comparison.set_index(
    "Experiment"
).plot(
    kind="bar",
    figsize=(10,6)
)

plt.title(
    "Original vs Augmented Dataset"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.tight_layout()

plt.savefig(
    "/Users/prakashpanta/Desktop/Cyber_Security_Risk_Assessment/Outputs/Figures/Original_vs_Augmented.png",
    dpi=300
)

plt.show()
```

## 3. Feature Removal Comparison

```python
comparison = df[
    df["Experiment"].isin(
        ["E1_Original_Full", "E2_Original_Reduced"]
    )
]

comparison.set_index(
    "Experiment"
).plot(
    kind="bar",
    figsize=(10,6)
)

plt.title(
    "Impact of Removing Training, MFA and Backup"
)

plt.ylabel(
    "Accuracy (%)"
)

plt.tight_layout()

plt.savefig(
    "/Users/prakashpanta/Desktop/Cyber_Security_Risk_Assessment/Outputs/Figures/Feature_Removal_Comparison.png",
    dpi=300
)

plt.show()
```
